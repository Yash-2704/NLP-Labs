# Experiment 2: Basic Text Preprocessing
##Experiment 2.1:  Write a Python program to perform basic text preprocessing on a given data. The program should:
1. Convert text to lowercase. (Also display how many uppercase?)
2. Remove punctuation.        (Also display how many different types of punctuations?)
3. Remove numbers.            (Also display how many numbers?)
4. Remove extra whitespace. (How much extra whitespace is present?)
5. Display the cleaned text.


```python
import string
import re

file_path = '/content/2.1_text_data.txt'

with open(file_path, 'r') as f:
    raw_text = f.read()

print(f"Original Text:\n{raw_text}")
```

### 1. Convert to Lowercase
We will count the number of uppercase characters before converting.

```python
upper_count = sum(1 for c in raw_text if c.isupper())
lowercase_text = raw_text.lower()

print(f"Number of uppercase characters: {upper_count}")
print(f"Lowercased Text:\n{lowercase_text}")
```

### 2. Remove Punctuation
We will identify the unique types of punctuation marks present.

```python
found_punct = [c for c in lowercase_text if c in string.punctuation]
unique_punct = set(found_punct)
no_punct_text = lowercase_text.translate(str.maketrans('', '', string.punctuation))

print(f"Number of different punctuation types: {len(unique_punct)} ({unique_punct})")
print(f"Text without Punctuation:\n{no_punct_text}")
```

### 3. Remove Numbers
We will count the total number of numeric digits removed.

```python
digit_count = sum(1 for c in no_punct_text if c.isdigit())
no_numbers_text = re.sub(r'\d+', '', no_punct_text)

print(f"Number of numeric digits found: {digit_count}")
print(f"Text without Numbers:\n{no_numbers_text}")
```

### 4. Remove Extra Whitespace
We will calculate the difference in length to see how much extra whitespace was removed.

```python
cleaned_text = " ".join(no_numbers_text.split())

whitespace_removed = len(no_numbers_text) - len(cleaned_text)
print(f"Characters of extra whitespace removed: {whitespace_removed}")
print(f"Final Cleaned Text:\n{cleaned_text}")
```

# Experiment 2.2: Tokenization
Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately using:
* NLTK
* spaCy
* Without using NLTK/spaCy (Pure Python)

```python
file_path = '/content/2.2_tokenization_data.txt'
with open(file_path, 'r') as f:
    text_data = f.read()
print("Data Loaded Successfully.")
```

### 1. Tokenization using NLTK

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize, sent_tokenize

nltk_sentences = sent_tokenize(text_data)
nltk_words = word_tokenize(text_data)

print("NLTK Sentences:", nltk_sentences)
print(f"Sentence Count: {len(nltk_sentences)}")
print("\nNLTK Words:", nltk_words)
print(f"Word Count: {len(nltk_words)}")
```

### 2. Tokenization using spaCy

```python
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp(text_data)

spacy_sentences = [sent.text for sent in doc.sents]
spacy_words = [token.text for token in doc]

print("spaCy Sentences:", spacy_sentences)
print(f"Sentence Count: {len(spacy_sentences)}")
print("\nspaCy Words:", spacy_words)
print(f"Word Count: {len(spacy_words)}")
```

### 3. Tokenization without NLTK/spaCy

```python
manual_sentences = text_data.replace('!', '.').replace('?', '.').split('.')
manual_sentences = [s.strip() for s in manual_sentences if s.strip()]

manual_words = text_data.split()

print("Manual Sentences:", manual_sentences)
print(f"Sentence Count: {len(manual_sentences)}")
print("\nManual Words:", manual_words)
print(f"Word Count: {len(manual_words)}")
```

# Experiment 2.3: Stop Word Removal
Write a Python program to tokenize a text and remove commonly occurring stop words. The program should:
* Display the original tokens.
* Display the filtered tokens (after removing stop words).
* Display the identified stop words separately.

```python
import re

file_path = '/content/2.3_clean_data.txt'
with open(file_path, 'r') as f:
    text_data = f.read()

original_tokens = re.findall(r'\w+', text_data.lower())

stop_words_list = ['is', 'the', 'a', 'an', 'in', 'of', 'and', 'to', 'it', 'for', 'on', 'with', 'as', 'this']

found_stop_words = [token for token in original_tokens if token in stop_words_list]
filtered_tokens = [token for token in original_tokens if token not in stop_words_list]

print(f"Original Tokens ({len(original_tokens)}):\n{original_tokens}\n")
print(f"Stop Words Found ({len(found_stop_words)}):\n{found_stop_words}\n")
print(f"Filtered Tokens ({len(filtered_tokens)}):\n{filtered_tokens}")
```
