# Experiment 3: Stemming, Lemmatization and Regular Expressions

## Experiment 3.1: Stemming using Porter Stemmer
Perform stemming on a list of words and display the original word alongside its stem.

```python
import nltk
nltk.download('punkt')
from nltk.stem import PorterStemmer

words = ["playing", "played", "plays", "studies", "studying",
         "connected", "connection", "computers"]

stemmer = PorterStemmer()

for word in words:
    print(f"{word} -> {stemmer.stem(word)}")
```

## Experiment 3.2: Lemmatization
Perform lemmatization on a list of words and display the original word alongside its lemma.

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
from nltk.stem import WordNetLemmatizer

words = ["cats", "dogs", "running", "runs", "ran", "studies", "studying",
         "better", "children", "mice", "went", "ate", "leaves", "caring"]

lemmatizer = WordNetLemmatizer()

for word in words:
    print(f"{word} -> {lemmatizer.lemmatize(word)}")
```

## Experiment 3.3: Extracting Information using Regular Expressions
Extract emails, URLs, mobile numbers, hashtags and mentions from a given text using regex.

```python
import re

text = """Welcome to the Natural Language Processing workshop! For more
information, visit https://www.nlpworkshop.com or www.python.org. You
can contact the coordinator at nlpworkshop@gmail.com or
support@python.org. For registration, call +91-9876543210 or 9123456789.
Follow us on social media @NLPWorkshop and @PythonLearner. Share your
experience using #NLP, #Python, and #MachineLearning. You can also visit
https://github.com/NLPWorkshop for the latest updates."""

emails = re.findall(r'[\w.+-]+@[\w-]+\.[\w.-]+', text)
urls = re.findall(r'https?://[^\s]+|www\.[^\s]+', text)
mobiles = re.findall(r'(?:\+91-?)?\d{10}', text)
hashtags = re.findall(r'#\w+', text)
mentions = re.findall(r'@\w+', text)

print("Emails:", emails)
print("URLs:", urls)
print("Mobile Numbers:", mobiles)
print("Hashtags:", hashtags)
print("Mentions:", mentions)
```
