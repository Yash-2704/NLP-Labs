### Experiment 4.1: Term-Frequency and NER (With Toolkit)
This experiment uses NLTK to calculate word frequencies and identify named entities.

```python
import nltk
import pandas as pd
from collections import Counter

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('maxent_ne_chunker')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

# Load the data
file_path = '/content/4.1_4.2_input.txt'
with open(file_path, 'r') as f:
  text = f.read()
# Term Frequency
words = nltk.word_tokenize(text.lower())
freq = Counter(words)

# Save to CSV
df_freq = pd.DataFrame(freq.items(), columns=['Term', 'Frequency'])
df_freq.to_csv('term_frequency_4_1.csv', index=False)

# Display Top 10
print("Top 10 Frequent Terms:")
display(df_freq.sort_values(by='Frequency', ascending=False).head(10))

# NER
tokens = nltk.word_tokenize(text)
tags = nltk.pos_tag(tokens)
entities = nltk.ne_chunk(tags)
print("\nNamed Entities:")
print(entities)
```

### Experiment 4.2: Term-Frequency Analysis (Without Toolkit)
Calculating frequency using standard Python dictionaries.

```python
import csv
import string

# Preprocess: remove punctuation and lowercase
clean_text = text.translate(str.maketrans('', '', string.punctuation)).lower()
words = clean_text.split()

# Calculate Frequency
freq_dict = {}
for word in words:
    freq_dict[word] = freq_dict.get(word, 0) + 1

# Save to CSV
with open('term_frequency_4_2.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['Term', 'Frequency'])
    for term, count in freq_dict.items():
        writer.writerow([term, count])

# Top 10
sorted_freq = sorted(freq_dict.items(), key=lambda x: x[1], reverse=True)
print("Top 10 Frequent Terms (Manual):")
for term, count in sorted_freq[:10]:
    print(f"{term}: {count}")
```

### Experiment 4.3: Manual TF-IDF Implementation
Calculating TF, DF, IDF, and TF-IDF for the provided documents.

```python
import math
import string

docs = [
    "Natural language processing is a field of artificial intelligence.",
    "Natural language processing helps computers understand human language.",
    "Machine learning is an important part of artificial intelligence."
]

def clean_and_tokenize(doc):
    doc = doc.lower().translate(str.maketrans('', '', string.punctuation))
    return doc.split()

tokenized_docs = [clean_and_tokenize(d) for d in docs]
all_terms = sorted(list(set([word for doc in tokenized_docs for word in doc])))

# 1. Calculate TF (Term Frequency)
tf_list = []
for doc in tokenized_docs:
    d_tf = {term: doc.count(term) / len(doc) for term in set(doc)}
    tf_list.append(d_tf)

# 2. Calculate DF (Document Frequency)
df_dict = {}
for term in all_terms:
    df_dict[term] = sum(1 for doc in tokenized_docs if term in doc)

# 3. Calculate IDF
num_docs = len(docs)
idf_dict = {term: math.log10(num_docs / df_dict[term]) for term in all_terms}

# 4. Calculate TF-IDF and Display Results
for i, doc_tf in enumerate(tf_list):
    print(f"\n--- Document {i+1} ---")
    doc_results = []
    for term in doc_tf:
        tfidf = doc_tf[term] * idf_dict[term]
        doc_results.append((term, doc_tf[term], df_dict[term], idf_dict[term], tfidf))

    # Display first few as example
    print(f"{'Term':<15} | {'TF':<8} | {'DF':<5} | {'IDF':<8} | {'TF-IDF'}")
    for res in sorted(doc_results, key=lambda x: x[4], reverse=True):
        print(f"{res[0]:<15} | {res[1]:.4f}   | {res[2]:<5} | {res[3]:.4f}   | {res[4]:.4f}")

    print(f"\nTop 10 Terms for Doc {i+1}:", [x[0] for x in sorted(doc_results, key=lambda x: x[4], reverse=True)[:10]])
```
