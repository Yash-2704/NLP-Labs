### Question 5.1: BPE Subword-level Tokenization

Implement subword-level tokenization using a BPE-based tokenizer.
Tokenize the sentence into subword units and display the generated tokens and their corresponding token IDs.

*   **a. Using a pretrained model**
*   **b. Without using a pretrained model**

```python
# Install necessary libraries
!pip install tokenizers transformers -q
```

```python
# 5.1.a Using Pretrained BPE Model (GPT-2 is BPE-based)
from transformers import GPT2Tokenizer

tokenizer_pretrained = GPT2Tokenizer.from_pretrained('gpt2')

# Read the entire input data file
with open('/content/input_sub_word_data.txt', 'r', encoding='utf-8') as f:
    full_text = f.read().strip()

# Note: the "sequence length > 1024" warning is harmless here, we only tokenize and do not run the model
tokens_pretrained = tokenizer_pretrained.tokenize(full_text)
token_ids_pretrained = tokenizer_pretrained.convert_tokens_to_ids(tokens_pretrained)

print("--- Pretrained GPT-2 BPE Tokenizer (complete input file) ---")
print("Total tokens:", len(tokens_pretrained))
print(f"{'Token':<25} | Token ID")
print("-" * 40)
for t, i in zip(tokens_pretrained, token_ids_pretrained):
    print(f"{t:<25} | {i}")
```

```python
# 5.1.b Train BPE from scratch (Without using a pretrained model)
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

# Initialize a BPE tokenizer
tok_bpe = Tokenizer(BPE(unk_token="[UNK]"))
tok_bpe.pre_tokenizer = Whitespace()

# vocab_size=300: small enough that BPE stops merging before whole words form,
# so reusable subword units (e.g. 'tion', 'ing', 'un') remain visible
trainer = BpeTrainer(special_tokens=["[UNK]", "[PAD]", "[CLS]", "[SEP]", "[MASK]"], vocab_size=300)
tok_bpe.train(["/content/input_sub_word_data.txt"], trainer)
print("Learned vocabulary size:", tok_bpe.get_vocab_size())

# Encode the entire file
encoded = tok_bpe.encode(full_text)

print("\n--- Trained from Scratch BPE Tokenizer (complete input file) ---")
print("Total tokens:", len(encoded.tokens))
print(f"{'Token':<25} | Token ID")
print("-" * 40)
for t, i in zip(encoded.tokens, encoded.ids):
    print(f"{t:<25} | {i}")

# Test on new words suggested in the input data
test_words = ["naturalization", "internationalization", "computationally", "unbelievable",
              "preprocessing", "classification", "communication", "renewable",
              "hydrokinetic", "multilingual", "understanding", "modernization"]
print("\n--- Subword segmentation of test words ---")
for w in test_words:
    e = tok_bpe.encode(w)
    print(f"{w:<22} -> {e.tokens}  IDs: {e.ids}")
```

### Question 5.2: SentencePiece Tokenization

Implement subword-level tokenization using a SentencePiece tokenizer with the same input file `/content/input_sub_word_data.txt`.
Tokenize the sentence into subword units and display the generated tokens and their corresponding token IDs.

*   **a. Using a pretrained model**
*   **b. Without using a pretrained model**

```python
# Install sentencepiece library
!pip install sentencepiece -q
```

```python
# 5.2.a Using Pretrained SentencePiece (ALBERT is SentencePiece-based)
from transformers import AlbertTokenizer

tokenizer_sp_pre = AlbertTokenizer.from_pretrained('albert-base-v2')

with open('/content/input_sub_word_data.txt', 'r', encoding='utf-8') as f:
    full_text = f.read().strip()

# Note: the "sequence length > 512" warning is harmless here, we only tokenize
tokens_sp_pre = tokenizer_sp_pre.tokenize(full_text)
token_ids_sp_pre = tokenizer_sp_pre.convert_tokens_to_ids(tokens_sp_pre)

print("--- Pretrained ALBERT SentencePiece Tokenizer (complete input file) ---")
print("Total tokens:", len(tokens_sp_pre))
print(f"{'Token':<25} | Token ID")
print("-" * 40)
for t, i in zip(tokens_sp_pre, token_ids_sp_pre):
    print(f"{t:<25} | {i}")
```

```python
# 5.2.b Train SentencePiece from scratch (Without using a pretrained model)
import sentencepiece as spm

# vocab_size=400: 100 was too small (output was almost character-level);
# 400 lets meaningful subword pieces form while staying within this corpus's limit
spm.SentencePieceTrainer.train(
    input='/content/input_sub_word_data.txt',
    model_prefix='sp_model',
    vocab_size=400,
    model_type='unigram',
    character_coverage=1.0
)

# Load trained model
sp = spm.SentencePieceProcessor(model_file='sp_model.model')

with open('/content/input_sub_word_data.txt', 'r', encoding='utf-8') as f:
    full_text = f.read().strip()

tokens_sp_scratch = sp.encode_as_pieces(full_text)
token_ids_sp_scratch = sp.encode_as_ids(full_text)

print("--- Trained from Scratch SentencePiece (complete input file) ---")
print("Total tokens:", len(tokens_sp_scratch))
print(f"{'Token':<25} | Token ID")
print("-" * 40)
for t, i in zip(tokens_sp_scratch, token_ids_sp_scratch):
    print(f"{t:<25} | {i}")

test_words = ["naturalization", "internationalization", "unbelievable", "hydrokinetic", "multilingual"]
print("\n--- Subword segmentation of test words ---")
for w in test_words:
    print(f"{w:<22} -> {sp.encode_as_pieces(w)}  IDs: {sp.encode_as_ids(w)}")
```

### Question 5.3: Part-of-Speech (POS) Tagging on Sample Text

Implement Part-of-Speech (POS) tagging on the text: `"The young student is reading an interesting book in the library."` and display the unique token, POS tag, and description.

*   **a. Using SpaCy**
*   **b. Using NLTK**

```python
# Install & download Spacy and NLTK resources
!python -m spacy download en_core_web_sm -q
import nltk
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('averaged_perceptron_tagger', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)
nltk.download('tagsets', quiet=True)
```

```python
# 5.3.a Using SpaCy
import spacy

nlp = spacy.load("en_core_web_sm")
text = "The young student is reading an interesting book in the library."
doc = nlp(text)

seen_tokens = set()
print(f"{'Token':<15} | {'POS Tag':<8} | {'Description'}")
print("-" * 60)
for token in doc:
    if token.text.lower() not in seen_tokens:
        seen_tokens.add(token.text.lower())
        print(f"{token.text:<15} | {token.pos_:<8} | {spacy.explain(token.pos_)}")
```

```python
# 5.3.b Using NLTK
import nltk
from nltk.tokenize import word_tokenize
from nltk.data import load

# Ensure required NLTK resources are downloaded
nltk.download('punkt_tab', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)
nltk.download('tagsets', quiet=True)
nltk.download('tagsets_json', quiet=True)

# Load official Penn Treebank tag descriptions (covers every tag, no hardcoding)
try:
    tagdict = load('help/tagsets_json/PY3_json/upenn_tagset.json')
except Exception:
    tagdict = load('help/tagsets/upenn_tagset.pickle')

text = "The young student is reading an interesting book in the library."
tokens = word_tokenize(text)
pos_tags = nltk.pos_tag(tokens)

seen_tokens = set()
print(f"{'Token':<15} | {'POS Tag':<8} | {'Description'}")
print("-" * 60)
for token, tag in pos_tags:
    if token.lower() not in seen_tokens:
        seen_tokens.add(token.lower())
        desc = tagdict.get(tag, ('unknown',))[0]
        print(f"{token:<15} | {tag:<8} | {desc}")
```

### Question 5.4: POS Tagging & Frequency using SpaCy

Implement Part-of-Speech (POS) tagging on the given file `/content/input_sub_word_data.txt` and display the unique token, POS tag, description, and frequency using SpaCy.

```python
from collections import Counter
import spacy

nlp = spacy.load("en_core_web_sm")

# Read all text from file
with open('/content/input_sub_word_data.txt', 'r', encoding='utf-8') as f:
    large_text = f.read()

doc = nlp(large_text)

# Frequency of each (token, POS tag) pair, so a word used with two different
# POS tags gets a separate, correct count for each
pair_freq = Counter((token.text.lower(), token.pos_) for token in doc if not token.is_space)

print(f"Total unique (token, POS) entries: {len(pair_freq)}\n")
print(f"{'Token':<22} | {'POS Tag':<8} | {'Frequency':<10} | {'Description'}")
print("-" * 75)

for (tok, pos), freq in pair_freq.most_common():
    print(f"{tok:<22} | {pos:<8} | {freq:<10} | {spacy.explain(pos)}")
```

### Question 5.5: Train Word2Vec CBOW Model

Create a file named `corpus.txt` with the given input corpus, train a Word2Vec CBOW model, and find the three most similar words to a selected word.

```python
# Create corpus.txt
corpus_content = """Natural language processing is interesting
Natural language processing is useful
I love machine learning
Machine learning is useful
Deep learning is a part of machine learning
Artificial intelligence is changing the world
Natural language processing uses machine learning
Deep learning and artificial intelligence are related
I study natural language processing
Machine learning helps artificial intelligence"""

with open('corpus.txt', 'w', encoding='utf-8') as f:
    f.write(corpus_content)
```

```python
# Install gensim library for Word2Vec
!pip install gensim -q
```

```python
from gensim.models import Word2Vec

# Load and preprocess corpus
sentences = []
with open('corpus.txt', 'r', encoding='utf-8') as f:
    for line in f:
        if line.strip():
            sentences.append(line.strip().lower().split())

# Hyperparameters tuned for this very small (10-sentence) corpus:
#   vector_size=20 : few words, so small vectors avoid too many free parameters
#   window=2       : sentences are short (4-8 words), a wide window mixes unrelated words
#   epochs=300     : few examples, so more passes are needed to learn anything
#                    (500+ over-trains: every word becomes ~0.98 similar to every other)

# Train CBOW model (sg=0 represents CBOW); seed + workers=1 make results reproducible
cbow_model = Word2Vec(sentences, vector_size=20, window=2, min_count=1, sg=0,
                      epochs=300, seed=42, workers=1)

# Selected word representation
selected_word = "learning"
print(f"Vector Representation of '{selected_word}':\n", cbow_model.wv[selected_word])

# Find three most similar words
similar_words = cbow_model.wv.most_similar(selected_word, topn=3)
print(f"\nThree most similar words to '{selected_word}':")
for w, s in similar_words:
    print(f"  {w:<15} {s:.4f}")
```

### Question 5.6: Train Word2Vec Skip-Gram Model

Train a Skip-Gram Word2Vec model using the same `corpus.txt`. Retrieve the vector for `"language"` and display the five most similar words.

```python
# Train Skip-Gram model (sg=1 represents Skip-Gram)
sg_model = Word2Vec(sentences, vector_size=20, window=2, min_count=1, sg=1,
                    epochs=300, seed=42, workers=1)

# Selected word representation
selected_word_sg = "language"
print(f"Vector Representation of '{selected_word_sg}':\n", sg_model.wv[selected_word_sg])

# Find five most similar words
similar_words_sg = sg_model.wv.most_similar(selected_word_sg, topn=5)
print(f"\nFive most similar words to '{selected_word_sg}':")
for w, s in similar_words_sg:
    print(f"  {w:<15} {s:.4f}")
```

### Question 5.7: Word2Vec Cosine Similarity

Train a Word2Vec model using the provided corpus and calculate the cosine similarity between selected word pairs:

*   machine and learning
*   natural and language
*   deep and learning

```python
# Train Word2Vec model
w2v_model = Word2Vec(sentences, vector_size=20, window=2, min_count=1, sg=1,
                     epochs=300, seed=42, workers=1)

pairs = [
    ('machine', 'learning'),
    ('natural', 'language'),
    ('deep', 'learning')
]

for w1, w2 in pairs:
    similarity = w2v_model.wv.similarity(w1, w2)
    print(f"Cosine similarity between '{w1}' and '{w2}': {similarity:.4f}")

# Baseline: an unrelated pair, to show the model separates related from unrelated words
print(f"\n(Baseline) 'language' vs 'world': {w2v_model.wv.similarity('language', 'world'):.4f}")
```

**Note on Word2Vec results (5.5–5.7):** The experiment specifies this 10-sentence `corpus.txt`, so it is used as given. Because the corpus is tiny, the hyperparameters were tuned for it (vector_size=20, window=2, epochs=300). The related pairs score about 0.7–0.9, well above the unrelated baseline (language–world, about 0.45). Common words such as 'a', 'is' and 'the' may still appear among the nearest neighbours, because in such a small corpus they share contexts with almost every word.
